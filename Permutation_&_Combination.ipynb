```python
def factorial(n):
    result = 1
    for i in range(1, n+ 1):
        result=result*i
    return result
n = int(input("Enter n: "))
r = int(input("Enter r: "))
n_fact=factorial(n)
formula=factorial(n - r)
r_fact=factorial(r)
npr =n_fact//formula
ncr = npr // r_fact
print("--------------------------")
print("Permutations :",npr)
print("--------------------------")
print("Combinations :", ncr)
```
